```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/HousePricePrediction.csv')
```

```python
print("First 10 Rows of Dataset:\n")
print(df.head(10))
```

#### Shape of dataset

```python
print("\nSummary Statistics:\n")
print(df.describe())
```

# =================================
# 4. Data Cleaning
# =================================



### Check missing values

```python
print("\nMissing Values:\n")
print(df.isnull().sum())
```

#### Handle missing values

```python
# Numerical columns -> fill with mean
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    df[col].fillna(df[col].mean(), inplace=True)

# Categorical columns -> fill with mode
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing values handled successfully.")

# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Remove duplicates if any
df.drop_duplicates(inplace=True)

print("Duplicates removed successfully.")


```

# =================================
# 5. Feature Selection
# =================================

```python
# Remove unnecessary column 'Id'
if 'Id' in df.columns:
    df.drop('Id', axis=1, inplace=True)

print("\nColumns after Feature Selection:\n")
print(df.columns)
```

# =================================
# 6. Data Preprocessing
# =================================

```python

# Convert categorical variables into numerical form
label_encoder = LabelEncoder()

for col in cat_cols:
    if col in df.columns:
        df[col] = label_encoder.fit_transform(df[col])

# Display processed dataset preview
print("\nPreprocessed Dataset Preview:\n")
print(df.head())

# Final dataset info
print("\nFinal Dataset Shape:")
print(df.shape)
```

# **📘 Hands-on Exercise 3: Machine Learning Model Building and Evaluation**

###In this exercise, I will build and evaluate a Machine Learning model using the cleaned dataset prepared in Hands-on Exercise 2.

I will begin by splitting the dataset into training and testing sets. Then, I will train a Linear Regression model to predict house prices based on the available features.

After training the model, I will use it to make predictions on the test data. I will evaluate the model using the following performance metrics:



*   Mean Absolute Error (MAE)
*   Mean Squared Error (MSE)
*   Root Mean Squared Error (RMSE)
*   R² Score


Finally, I will analyze the results to determine how well the model is performing and whether it needs further improvement.

The objective of this exercise is to help me understand the complete machine learning workflow, from data preparation to model evaluation.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

#### **Define Features and Target**

```python
X = df.drop('SalePrice', axis=1)
y = df['SalePrice']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

#### **Train-Test Split (80% Training, 20% Testing)**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

### **Train Linear Regression Model**

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Model trained successfully.")
```

#### **Predict House Prices**

```python
# Predict on test data
y_pred = model.predict(X_test)

print("Predicted House Prices:\n")
print(y_pred[:10])
```

### **Model Evaluation**

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

# Print results
print("Model Evaluation Results:\n")

print("Mean Absolute Error (MAE):", mae)

print("Mean Squared Error (MSE):", mse)

print("Root Mean Squared Error (RMSE):", rmse)

print("R² Score:", r2)
```

### **Interpretation**

```python
print("\nModel Interpretation:\n")

if r2 > 0.80:
    print("The model performance is very good.")
elif r2 > 0.60:
    print("The model performance is acceptable but can be improved.")
else:
    print("The model performance is weak and needs improvement.")
```

## **Interpretation of Results**



*   The Linear Regression model was successfully trained using the house price dataset.
*   The dataset was divided into training and testing sets using an 80:20 ratio.

*   The Linear Regression model was evaluated using MAE, MSE, RMSE, and R² Score.
*   The Mean Absolute Error (MAE = 32540.54) indicates that, on average, the model’s predictions are off by around 32,540 units of house price, which is relatively high.
*   The Root Mean Squared Error (RMSE = 48706.09) is also high, showing that the model has significant prediction errors, especially for larger values.
*   The R² Score = 0.3349 means that the model is only able to explain about 33% of the variance in house prices, which is considered low performance.

## **⚠️ Conclusion:**

Overall, the model performance is weak. This indicates that Linear Regression is not capturing the relationship between features and house prices effectively.
