# **1️⃣ Scalars**

## A scalar is a single numerical value.

# **✅ Example**

```python
scalar = np.array(10)

print("Scalar:", scalar)
print("Dimensions:", scalar.ndim)
print("Shape:", scalar.shape)
```

# **2️⃣ Vectors**

A vector is an ordered list of numbers.

✅ Example

```python
vector = np.array([2, 4, 6])

print("Vector:", vector)
print("Dimensions:", vector.ndim)
print("Shape:", vector.shape)
```

# 3️⃣ Matrix

A matrix contains rows and columns.

## ✅ Example

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A)
print("Shape:", A.shape)
print("Dimensions:", A.ndim)
```

## 📌 Machine Learning Dataset as Matrix

```python
X = np.array([
    [21, 5, 3.7],
    [22, 6, 3.5],
    [20, 4, 3.9]
])

print(X)
```

## 📊 Understanding Shape

```python
print("Rows (samples):", X.shape[0])
print("Columns (features):", X.shape[1])
```
