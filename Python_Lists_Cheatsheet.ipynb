### 1. Creating a List

Lists are ordered, mutable collections that can store mixed data types and allow duplicates. Their size is dynamic.

```python
empty_list = []
nums = [1, 2, 3]
mixed = [1, "hello", 3.14, True]
nested = [1, [2, 3], {"a": 1}]
```

### 2. Accessing Elements

Elements can be accessed by index (both positive and negative for reverse indexing) and by slicing.

```python
lst = ['a', 'b', 'c', 'd', 'e']

print(lst[0])      # Access first element
print(lst[-1])     # Access last element
print(lst[1:4])    # Slice from index 1 (inclusive) to 4 (exclusive)
print(lst[::-1])   # Create a reverse copy of the list
```

### 3. Modifying Lists

Individual elements or slices of a list can be modified by assignment.

```python
lst = ['a', 'b', 'c', 'd', 'e'] # Re-initialize lst for demonstration
lst[1] = 'B' # Modify a single element
lst[1:3] = ['X', 'Y'] # Modify a slice
print(lst)
```

### 4. Adding Elements

Elements can be added to a list using `append()`, `insert()`, or `extend()`.

```python
lst = [1, 2, 3] # Re-initialize lst for demonstration
lst.append('Z') # Adds element to the end
lst.insert(1, 'NEW') # Inserts element at a specific index
lst.extend([10, 20]) # Extends list by appending elements from an iterable
print(lst)
```

### 5. Removing Elements

Elements can be removed using `remove()`, `pop()`, or `del`.

```python
lst = ['A', 'B', 'C', 'D', 'E'] # Re-initialize lst for demonstration
lst.remove('C') # Removes the first occurrence of a value
last = lst.pop() # Removes and returns the last element (or by index)
del lst[0] # Deletes an element by index
print(lst)
print(f"Popped element: {last}")
```

### 6. Searching

Check for element presence using the `in` operator or count occurrences with `count()`.

```python
lst = ['apple', 'banana', 'cherry', 'banana']
print('banana' in lst) # Check if 'banana' is in the list
print(lst.count('banana')) # Count occurrences of 'banana'
```

### 7. List Operations

Common operations include getting length, sorting (in-place or returning a new list), and reversing.

```python
nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(len(nums)) # Get the number of elements
print(sorted(nums)) # Returns a new sorted list, original remains unchanged

nums_copy = list(nums) # Create a copy to show in-place sort
nums_copy.sort() # Sorts the list in-place
print(nums_copy)

nums_copy.reverse() # Reverses the list in-place
print(nums_copy)
```

### 8. Nested Lists

Lists can contain other lists, allowing for multi-dimensional data structures like matrices.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0][1]) # Access element at row 0, column 1
```

### 9. Iteration

Lists can be iterated using `for` loops. `enumerate()` provides both index and value.

```python
nums = [10, 20, 30]
for item in nums:
    print(f"Value: {item}")

for i, item in enumerate(nums):
    print(f"Index: {i}, Value: {item}")
```

### 10. List Comprehensions

A concise way to create lists. They can include expressions, `for` loops, and `if` conditions.

```python
squares = [x**2 for x in range(5)] # Create a list of squares
evens = [x for x in range(10) if x % 2 == 0] # Create a list of even numbers
matrix2 = [[i * j for j in range(3)] for i in range(3)] # Nested list comprehension

print(squares)
print(evens)
print(matrix2)
```

### 11. Built-in Functions

Python offers several built-in functions that work with lists.

```python
data = [1, 2, 3, 4, 5]
print(sum(data)) # Sum of all elements
print(min(data)) # Smallest element
print(max(data)) # Largest element
print(any([0, 0, 1])) # True if any element is truthy
print(all([1, 1, 1])) # True if all elements are truthy
print(all([1, 0, 1])) # False because of 0
```

### 12. Time Complexity (Average)

Understanding the average time complexity of common list operations is crucial for performance.

- **Access**: O(1) - Constant time to access an element by index.
- **Append**: O(1) - Amortized constant time to add an element to the end.
- **Insert/Delete middle**: O(n) - Linear time as elements need to be shifted.
- **Search**: O(n) - Linear time to find an element.
- **Sort**: O(n log n) - Log-linear time for typical sorting algorithms.

### More Common Operations

These are additional frequently used methods for list manipulation.

```python
nums = [1, 2, 3, 1] # Initialize a list for demonstration

print(f"Original list: {nums}")
print(f"Index of first '1': {nums.index(1)})")      # Find index of a value

nums.append(5)     # Add to the end of the list
print(f"After append(5): {nums}")

nums.insert(0, 10)  # Insert 10 at index 0
print(f"After insert(0, 10): {nums}")

nums.remove(3)     # Remove the first occurrence of value 3
print(f"After remove(3): {nums}")

popped_item = nums.pop()         # Remove and return the last element
print(f"After pop(), popped item: {popped_item}, list: {nums}")

nums.sort()        # Sort the list in-place (TimSort: O(n log n))
print(f"After sort(): {nums}")

nums.reverse()     # Reverse the list in-place
print(f"After reverse(): {nums}")

shallow_copy = nums.copy()        # Return a shallow copy of the list
print(f"Shallow copy: {shallow_copy}")
```

### List Slicing

Slicing provides a powerful way to extract sub-lists or modify parts of a list.

```python
nums = [10, 20, 30, 40, 50, 60]
print(f"Original list for slicing: {nums}")

print(f"nums[1:4] (elements from index 1 up to but not including 4): {nums[1:4]}")  # Generic slice syntax
print(f"nums[-1] (last item): {nums[-1]}")    # Last item
print(f"nums[::-1] (reverse list): {nums[::-1]}")  # Reverse list
print(f"nums[1:] (everything after index 1): {nums[1:]}")    # Everything after index 1
print(f"nums[:3] (first three elements): {nums[:3]}")    # First three elements
print(f"nums[::2] (every other element): {nums[::2]}")    # Every other element
```

### Quick Example: List Manipulation Flow

This example demonstrates a sequence of common list manipulations.

```python
lst = [3, 1, 4] # Initial list
lst.append(2) # Add 2 to the end: [3, 1, 4, 2]
lst.insert(1, 9) # Insert 9 at index 1: [3, 9, 1, 4, 2]
lst.remove(4) # Remove the value 4: [3, 9, 1, 2]
lst[0] = 7 # Change element at index 0: [7, 9, 1, 2]
last = lst.pop() # Remove and get last element (2): [7, 9, 1]
lst.sort() # Sort the list in-place: [1, 7, 9]

print(lst) # Final state of the list
```
