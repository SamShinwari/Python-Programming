#LIST

**What is a List?**

*A list in Python is a collection of items stored in a single variable.*

*   Ordered
*   Mutable
*   Allows duplicate values


```python
my_list = [1, 2, 3, 4, 5]
print(my_list)
```

📌 Creating Lists

```python
empty_list = []

# List with different data types
mixed_list = [1, "Hello", 3.5, True]

# Nested list
nested_list = [[1, 2], [3, 4]]


print(mixed_list)
print(nested_list)
```

📌 Accessing List Elements

```python
my_list = [10, 20, 30, 40]

# Indexing
print(my_list[0])   # First element
print(my_list[-1])  # Last element

# Slicing
print(my_list[1:3])  # [20, 30]
```

📌 List Methods

```python

my_list = [1, 2, 3]
print(my_list)

# Add elements
my_list.append(4)
print("list after append method",my_list)

my_list.insert(1, 10)
print("list after insert method",my_list)


# Remove elements
my_list.remove(2)
print("list after remove method",my_list)

my_list.pop()
print("list after pop method",my_list)


# Sort list
my_list.sort()
print("list after sort method",my_list)

# Reverse list
my_list.reverse()
print("list after reverse method",my_list)
print(my_list)

```

📌 List Operations


```python
list1 = [1, 2, 3]
list2 = [4, 5]

# Concatenation
print(list1 + list2)

# Repetition
print(list1 * 2)

# Membership
print(2 in list1)  # True
```

📌 Looping Through a List

```python
my_list = [1, 2, 3, 4]

for item in my_list:
    print(item)
```

📌 List Comprehension (Important 🔥)


```python
# Create list of squares
squares = [x**2 for x in range(5)]
print(squares)
```

📌 Useful Functions


```python
numbers = [4, 2, 8, 1]

print(len(numbers))   # Length
print(max(numbers))   # Maximum
print(min(numbers))   # Minimum
print(sum(numbers))
```
